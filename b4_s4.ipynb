```python
#column cleanup

df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(r"[^a-z0-9]+", "_", regex=True)
      .str.strip("_")
)

print(df.columns.tolist())
```

```python
df["student_id"]=df["student_id"].astype("string")
df["city"]=df["city"].astype("category")
```

```python
#MISSING VALUES

missing_demo = df.copy()

missing_demo.loc[2, "science"] = np.nan
missing_demo.loc[8, "attendance_percent"] = np.nan
missing_demo.loc[14, "city"] = None

print(missing_demo.isna().sum())
```

```python
science_median= missing_demo["science"].median()
attendance_median = missing_demo["attendance_percent"].median()
city_mode= missing_demo["city"].mode()[0]


missing_demo["science"] = missing_demo["science"].fillna(science_median)
```

```python
#averages and some basic summaries
#derive a new column (feature creation) for each student we will calculate their subject averages

df["average_score"]= (
    df[["math", "science", "english"]]
    .mean(axis=1)
    .round(2)
)

print(df[["name","math","science","english","average_score"]].head())

```

```python
print("Class average:", df["average_score"].mean())
print("Lowest average:", df["average_score"].min())
print("Highest average:", df["average_score"].max())
print("Students:", df["student_id"].count())

print(df['average_score'].describe())
```

```python
#count mean, minumum and max

#loc means label based indexing
top_student = df.loc[df["average_score"].idxmax(), ["name", "average_score"]]
print(top_student)

low_attendance= df[df["attendance_percent"]<75]
print(low_attendance[["name", "attendance_percent", "average_score"]])
#
```
