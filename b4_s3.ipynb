```python
def add_bonus_in_place(marks, bonus):
    for index in range(len(marks)):
        marks[index] = min(100, marks[index] + bonus)


def with_bonus(marks, bonus):
    return [min(100, mark + bonus) for mark in marks]

original_marks = [72, 81, 90]
updated_marks = with_bonus(original_marks, 5)

print("original:", original_marks)
print("updated:", updated_marks)
```
